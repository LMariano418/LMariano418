```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = df_from_drive

print("Primeiras linhas do dataset:")
print(df.head())

print("\nInformações gerais:")
print(df.info())

print("\nEstatísticas descritivas:")
print(df.describe())

col_preco = 'Preço'
col_nota = 'Nota'
col_categoria = 'Temporada'

plt.figure()
plt.hist(df[col_preco].dropna(), bins=30)
plt.title('Histograma - Distribuição dos Preços')
plt.xlabel('Preço')
plt.ylabel('Frequência')
plt.show()

plt.figure()
plt.scatter(df[col_preco], df[col_nota])
plt.title('Dispersão - Preço vs Nota')
plt.xlabel('Preço')
plt.ylabel('Nota')
plt.show()

plt.figure()
correlacao = df.corr(numeric_only=True)

plt.imshow(correlacao)
plt.title('Mapa de Calor - Correlação entre Variáveis Numéricas')
plt.xticks(range(len(correlacao.columns)), correlacao.columns, rotation=90)
plt.yticks(range(len(correlacao.columns)), correlacao.columns)
plt.colorbar()
plt.show()

plt.figure()
df[col_categoria].value_counts().plot(kind='bar')
plt.title('Quantidade de Produtos por Temporada')
plt.xlabel('Temporada')
plt.ylabel('Quantidade')
plt.show()

plt.figure()
df[col_categoria].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Participação das Temporadas')
plt.ylabel('')
plt.show()

plt.figure()
df[col_preco].dropna().plot(kind='density')
plt.title('Densidade - Distribuição dos Preços')
plt.xlabel('Preço')
plt.show()

df_reg = df[[col_preco, col_nota]].dropna()

x = df_reg[col_preco]
y = df_reg[col_nota]

coef = np.polyfit(x, y, 1)
polinomio = np.poly1d(coef)

plt.figure()
plt.scatter(x, y)
plt.plot(x, polinomio(x))
plt.title('Regressão Linear - Preço vs Nota')
plt.xlabel('Preço')
plt.ylabel('Nota')
plt.show()
```
