# 1 - Carregue a base, verifique se os tipos de dados estão de acordo com cada coluna e caso não estejam realize a transformação dos tipos de dados.

```python
#seu código aqui
# estrutura
df.info()

# Ajuste

# Idoso como categoria
df['Idoso'] = df['Idoso'].astype('category')

# Colunas categóricas
cat_cols = [
    'Genero', 'Casado', 'Dependents', 'PhoneService',
        'Servico_Internet', 'Servico_Seguranca', 'Suporte_Tecnico',
            'StreamingTV', 'Tipo_Contrato', 'PaymentMethod', 'Churn'
            ]

for col in cat_cols:
    df[col] = df[col].astype('category')

# Numéricas
num_cols = ['Tempo_como_Cliente', 'Pagamento_Mensal', 'Total_Pago']
df[num_cols] = df[num_cols].apply(pd.to_numeric, errors='coerce')

df.info()
```

# 2 - Esse exercício faremos por etapas:

```python
#seu código aqui
missing = df.isnull().sum()
missing = missing.sort_values(ascending=False)

print(missing)
```

```python
# Preenchendo 'Pagamento_Mensal' com a mediana
# Verificar a distribuição antes de decidir (já foi feita anteriormente, agora só imputar)
df['Pagamento_Mensal'].fillna(df['Pagamento_Mensal'].median(), inplace=True)
```

```python
# Preenchendo 'PhoneService' com a moda
df['PhoneService'].fillna(df['PhoneService'].mode()[0], inplace=True)
```

# 3 - Verifique se encontramos valores digitados incorretamente, ou com letras maiusculas ou minusculas, ou até mesmo mesmos valores porém escritos de forma dirente e corrija.

```python
#seu código aqui

# padronizar texto
df['Servico_Internet'] = df['Servico_Internet'].str.lower().str.strip()
df['Genero'] = df['Genero'].str.lower().str.strip()

# corrigir genero
df['Genero'] = df['Genero'].replace({
    'f': 'female',
    'm': 'male'
})

# padronizar categorias gerais
for col in cat_cols:
    df[col] = df[col].astype(str).str.strip().str.lower()
```
